# Step 1 :Load files

```python
from pathlib import Path
import yaml
import pandas as pd

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "config.yaml").exists())

with open(ROOT / "config.yaml") as f:
    cfg = yaml.safe_load(f)

RAW = ROOT / cfg["paths"]["raw"]

sales = pd.read_csv(RAW / "sales_train_evaluation.csv")

calendar = pd.read_csv(RAW / "calendar.csv")

sell_prices = pd.read_csv(RAW / "sell_prices.csv")
```

```python
print("Shape of all files")
print("Sales shape", sales.shape)
print("Calendar shape", calendar.shape)
print("Sell price shape", sell_prices.shape)
```

### Scope 

- Filter only CA X FOODS 
- The below data in only of the Sales in CA for Food Category

```python
sales_ca = sales[(sales['state_id']==cfg["state"]) &  (sales["cat_id"]==cfg["category"])].copy()

print(sales_ca.head())
print(sales_ca.shape)
```

### Filter prices for the above table from "sell_prices" table

```python
# take unique Stores and Items from sales_ca Table

unique_pairs = sales_ca[['store_id', 'item_id']].drop_duplicates()

print(unique_pairs.head())
print(unique_pairs.shape)
```

```python
prices_ca = pd.merge(sell_prices, unique_pairs, on=['store_id', 'item_id'])

print(prices_ca.head())
print(prices_ca.shape)
```

```python
print(prices_ca["store_id"].unique())
print(prices_ca["item_id"].nunique())
```

```python
del sales, sell_prices
```

### Schema and keys

```python
# Count Null values in "sales_ca, calendar, prices_ca"

print("Null values in Sales of CA table",sales_ca.isna().sum().sum())

print("Null values in Calendar table", calendar.isna().sum())

print("Null values in prices of CA table", prices_ca.isna().sum().sum())


# the 7542 Nulls from calendar are all  from the 4 event columns: 1,807 + 1,807 + 1,964 + 1,964 = 7,542.
# They're expected, because a null there means "no event that day". It's not missing data.
```

```python
#   Check for duplicates

# Check if any duplicate ID in 
print("Duplicate row IDs in Sales CA :",sales_ca['id'].duplicated().sum())

print("Duplicate Days in Calendar table", calendar['d'].duplicated().sum())

print("Duplicates in Prices table", prices_ca.duplicated(subset=['store_id', 'item_id', 'wm_yr_wk']).sum())
```

### Process Calendar table

```python
# convert "date" column to Datetime Dtype

calendar['date'] = pd.to_datetime(calendar['date'])
first_date = calendar['date'].iloc[0]
last_date  = calendar['date'].iloc[-1]

print("Start date", first_date)
print("Last date", last_date)
```

```python
# Count the event days per event type.

event_1 = calendar['event_name_1'].value_counts()
event_2 = calendar['event_name_2'].value_counts()

print(event_1)
```
