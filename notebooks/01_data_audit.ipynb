# Step 1 :Load files

```python
from pathlib import Path
import yaml
import pandas as pd

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "config.yaml").exists())

with open(ROOT / "config.yaml") as f:
    cfg = yaml.safe_load(f)

RAW = ROOT / cfg["paths"]["raw"]

sales = pd.read_csv(RAW / "sales_train_evaluation.csv")

calendar = pd.read_csv(RAW / "calendar.csv")

sell_prices = pd.read_csv(RAW / "sell_prices.csv")
```

```python
print("Shape of all files")
print("Sales shape", sales.shape)
print("Calendar shape", calendar.shape)
print("Sell price shape", sell_prices.shape)
```

### Scope 

- Filter only CA X FOODS 
- The below data in only of the Sales in CA for Food Category

```python
sales_ca = sales[(sales['state_id']==cfg["state"]) &  (sales["cat_id"]==cfg["category"])].copy()

sales_ca.head()
```

### Filter prices for the above table from "sell_prices" table

```python
# take unique Stores and Items from sales_ca Table

unique_pairs = sales_ca[['store_id', 'item_id']].drop_duplicates()

unique_pairs.head()
```

```python
prices_ca = pd.merge(sell_prices, unique_pairs, on=['store_id', 'item_id'])

prices_ca.head()
```
