### Pre-launch zero sales

- For each row of sales_ca, find the first day with sales above 0

```python
# get where the sales > 0 for day columns

sold = sales_ca[day_cols] > 0


# get the first sale of each row
first_sale = sold.idxmax(axis=1)

first_sale.value_counts().head()


# In data: across the 5,748 rows, 1,763 rows had d_1 as their answer, 267 rows had d_2, and so on.
```

### Finding: pre-launch zeros
- For each item-store series, I found the first day with sales > 0.
- Only 1,763 of 5,748 series (~31%) were selling from day 1 (2011-01-29).
- Series that start after a few days (d_2–d_4) are likely slow sellers, already in store.
- Series that start much later (e.g. 60 series on d_1579, ~May 2015) are new product launches.
- **Impact:** zeros before a product's launch are not real "no demand"; they mean the product
  was not in the store. Training on them would teach the model false zero demand.
  → Input for decision #7 (handling pre-launch zeros).


### what share of all zeros are pre-launch?

```python
print(sold.any(axis=1).sum()) # check atleast 1 product to be sold across the data

# Mark the days before each row first sale
launched = sold.cummax(axis=1)  # once the product is on shelf it is marked in the store

# pre-launched will mark the days before the product arrives
pre_launched = ~launched

# filp the sold to mark zero days and count them
total_zeros = (~sold).sum().sum()

# Count the zeros in pre launched
pre_launch_zeros = pre_launched.sum().sum()

print("Pre-launch share of zeros:", pre_launch_zeros / total_zeros * 100)
```

### Finding: share of fake zeros
- 38% of all zero-sales days fall **before the product's first sale**

- These zeros mean "not in store yet", not "no demand".

----------------------------------------------------

- Dropping each series days before its first sale
- WHY?: Keeping them would teach model false zero demand
